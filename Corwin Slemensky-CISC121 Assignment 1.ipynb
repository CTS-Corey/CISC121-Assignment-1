```python
"""Problem 1: Cyclic Rotation"""
def solution(A, K):
    for n in range(K): #repeat K times
        temp = A[len(A)-1] # store last value in A
        for i in range(len(A)-1, 0, -1): # for each element 
            A[i] = A[i-1] # replace current value with previous value
        A[0] = temp # reset first value as last value
    return A # return array

print(solution([3,8,9,7,6], 3))
```

```python
"""Problem 2: Odd Occurrences in Array"""
def solution(A):
    unpaired = 0 # initialize unpaired as 0
    for number in A: # loop through A
        unpaired ^= number #check if value has a partner
    return unpaired # return unpaired value

print(solution([9,3,9,3,9,7,9]))
```

```python
"""Problem 3: Perm Missing Elem"""
def solution(A):
    N = len(A) # initialize N as length of A
    expected_sum = (N + 1) * (N + 2) // 2 # calculate expected sum
    actual_sum = sum(A) # calculate actual sum
    missing_number = expected_sum - actual_sum # calculate missing number
    return missing_number # return missing number

print(solution([2, 3, 1, 5]))
```

```python
"""Problem 4: Tape Equilibrium"""
def solution(A):
    total_sum = sum(A) # set total sum of A
    left_sum = 0 # initialize left sum
    min_difference = float('inf') # initialize min dif as infinity
    for P in range(1, len(A)): # loop through array
        left_sum += A[P - 1] # calculate left sum
        right_sum = total_sum - left_sum # calculate right sum
        difference = abs(left_sum - right_sum) # calculate difference
        if difference < min_difference: # if difference smaller than min dif
            min_difference = difference # set min dif to new difference
    return min_difference # return min dif

print(solution([3,1,2,4,3]))
```

```python
"""Problem 5: Permutation Check"""
def solution(A):
    N = len(A) # set N as length of A
    unique_elements = set() # initialize unique elements set
    expected_sum = N * (N + 1) // 2 # calculate expected sum
    actual_sum = 0 # calculate actual sum
    for number in A: # loop through A
        if number < 1 or number > N or number in unique_elements: # check if makes permutation invalid
            return 0 # return 0 (not a permutation)
        unique_elements.add(number) # add number to unique elements list
        actual_sum += number # add number to total sum
    if actual_sum == expected_sum: # if actual sum = expected sum
        return 1 # return 1 (is a permutation)
    else: # if not
        return 0 # return 0 (not a permutation)

print(solution([4, 1, 3, 2]))
print(solution([4, 1, 3]))
```
